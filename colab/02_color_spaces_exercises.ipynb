Practice converting between RGB, HSV, and HSL color spaces.

````python
# Setup
import math

def rgb_to_hsv(r: int, g: int, b: int) -> tuple[float, float, float]:
    rn, gn, bn = r / 255, g / 255, b / 255
    mx = max(rn, gn, bn)
    mn = min(rn, gn, bn)
    diff = mx - mn
    
    if diff < 0.001:
        h = 0.0
    elif mx == rn:
        h = 60 * (((gn - bn) / diff) % 6)
    elif mx == gn:
        h = 60 * (((bn - rn) / diff) + 2)
    else:
        h = 60 * (((rn - gn) / diff) + 4)
    if h < 0:
        h += 360
    
    s = diff / mx if mx > 0 else 0
    v = mx
    return (h, s, v)

def hsv_to_rgb(h: float, s: float, v: float) -> tuple[int, int, int]:
    c = v * s
    x = c * (1 - abs((h / 60) % 2 - 1))
    m = v - c
    
    if h < 60:
        rn, gn, bn = c, x, 0
    elif h < 120:
        rn, gn, bn = x, c, 0
    elif h < 180:
        rn, gn, bn = 0, c, x
    elif h < 240:
        rn, gn, bn = 0, x, c
    elif h < 300:
        rn, gn, bn = x, 0, c
    else:
        rn, gn, bn = c, 0, x
    
    return (int((rn + m) * 255), int((gn + m) * 255), int((bn + m) * 255))

print('Color functions loaded!')
````

### Exercise 1: RGB to HSV

Convert RGB(255, 0, 0) to HSV. What do you expect?
Then try RGB(0, 255, 0) and RGB(0, 0, 255).

````python
colors = [(255, 0, 0), (0, 255, 0), (0, 0, 255), (255, 255, 0), (128, 128, 128)]
for r, g, b in colors:
    h, s, v = rgb_to_hsv(r, g, b)
    print(f"RGB({r:3d}, {g:3d}, {b:3d}) -> HSV({h:.0f}°, {s:.2f}, {v:.2f})")
````

### Exercise 2: HSV to RGB

Convert HSV(0, 1, 1) and HSV(120, 1, 1) and HSV(240, 1, 1) to RGB.
What colors do you get?

````python
hsv_colors = [(0, 1, 1), (120, 1, 1), (240, 1, 1), (60, 0.5, 1), (300, 1, 0.5)]
for h, s, v in hsv_colors:
    r, g, b = hsv_to_rgb(h, s, v)
    print(f"HSV({h:.0f}°, {s:.2f}, {v:.2f}) -> RGB({r:3d}, {g:3d}, {b:3d})")
````
